```python
import pandas as pd

metrics = pd.read_parquet("../results/parquets/metrics_ridge_allfeats_delta.parquet")
preds = pd.read_parquet("../results/parquets/predictions_ridge_allfeats_delta.parquet")
```

```python
metrics.to_csv("../results/csvs/metrics_ridge_allfeats_delta.csv", index=False)
preds.to_csv("../results/csvs/predictions_ridge_allfeats_delta.csv", index=False)
```

```python
def for_excel(df):
    out = df.copy()
    for col in out.columns:
        if isinstance(out[col].dtype, pd.DatetimeTZDtype):
            out[col] = out[col].dt.tz_localize(None)
    return out


with pd.ExcelWriter("../results/excels/ridge_allfeats_delta.xlsx") as writer:
    for_excel(metrics).to_excel(writer, sheet_name="metrics", index=False)
    for_excel(preds).to_excel(writer, sheet_name="predictions", index=False)
```
