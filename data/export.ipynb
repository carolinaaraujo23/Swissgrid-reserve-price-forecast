```python
import pandas as pd

offers = pd.read_parquet("../data/raw/external")
offers.head()
```

```python
offers.to_csv("../data/processed/ext_day_ahead.csv", index=False)
```

```python
to_excel = offers.copy()

for col in to_excel.columns:
    if isinstance(to_excel[col].dtype, pd.DatetimeTZDtype):
        to_excel[col] = to_excel[col].dt.tz_localize(None)

to_excel.to_excel("../data/processed/ext_day_ahead.xlsx", index=False)
```
